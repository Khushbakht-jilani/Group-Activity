```python
def most_frequent_char(s):# create function
    freq = {}

    for char in s:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

    max_freq = 0
    result_char = ''

    for char in freq:
        if freq[char] > max_freq or (freq[char] == max_freq and char < result_char):
            max_freq = freq[char]
            result_char = char

    return result_char

# User input
user_input = input("Enter a string: ")
result = most_frequent_char(user_input)
print("Most frequent character:", result)

```

```python
#task 11

def print_primes_less_than_n(n):
    for num in range(2, n):
        is_prime = True
        for i in range(2, int(num**0.5)+1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num, end=', ')

# Example usage
n = int(input("Enter a number: "))
print_primes_less_than_n(n)

```

```python
def sum_all_numbers(*args):
    return sum(args)

# User input
user_input = input("Enter numbers separated by spaces: ")
numbers = tuple(map(int, user_input.split()))

# Function call
result = sum_all_numbers(*numbers)
print("Sum:", result)
```
