```python
def most_frequent_char(s):# create function
    freq = {}

    for char in s:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

    max_freq = 0
    result_char = ''

    for char in freq:
        if freq[char] > max_freq or (freq[char] == max_freq and char < result_char):
            max_freq = freq[char]
            result_char = char

    return result_char

# User input
user_input = input("Enter a string: ")
result = most_frequent_char(user_input)
print("Most frequent character:", result)

```

```python
#task 11

def print_primes_less_than_n(n):
    for num in range(2, n):
        is_prime = True
        for i in range(2, int(num**0.5)+1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num, end=', ')

# Example usage
n = int(input("Enter a number: "))
print_primes_less_than_n(n)

```

```python
#task1
def largest_difference(arr):
    if not arr or len(arr) < 2:
        return 0

    min_element = arr[0]
    max_diff = 0

    for i in range(1, len(arr)):
        if arr[i] > min_element:
            max_diff = max(max_diff, arr[i] - min_element)
        else:
            min_element = arr[i]

    return max_diff

# User input
user_input = input("Enter a list of integers separated by spaces: ")
numbers = list(map(int, user_input.strip().split()))
result = largest_difference(numbers)
print("The largest difference is:", result)
```

```python
#task2
def first_non_repeating_char(s):
    char_count = {}

    # Count each character
    for char in s:
        char_count[char] = char_count.get(char, 0) + 1

    # Find first character with count 1
    for char in s:
        if char_count[char] == 1:
            return char

    return None

# User input
user_input = input("Enter a string: ")
result = first_non_repeating_char(user_input)
if result:
    print("The first non-repeating character is:", result)
else:
    print("There are no non-repeating characters.")
```

```python
#task3
from collections import Counter

def are_permutations(str1, str2):
    return Counter(str1) == Counter(str2)

# User input
s1 = input("Enter the first string: ")
s2 = input("Enter the second string: ")

if are_permutations(s1, s2):
    print("True – The strings are permutations of each other.")
else:
    print("False – The strings are NOT permutations of each other.")
```

```python
#task4
def second_largest_unique(nums):
    unique_nums = list(set(nums))

    if len(unique_nums) < 2:
        return None

    unique_nums.sort(reverse=True)
    return unique_nums[1]

# User input
user_input = input("Enter a list of integers separated by spaces: ")
numbers = list(map(int, user_input.strip().split()))
result = second_largest_unique(numbers)

if result is not None:
    print("The second largest unique number is:", result)
else:
    print("There is no second largest unique number.")
```

```python
#task5
def count_pairs_with_sum(nums, target):
    seen = set()
    pairs = set()

    for num in nums:
        complement = target - num
        if complement in seen:
            # Store pair in sorted order to avoid duplicates like (2,3) and (3,2)
            pairs.add(tuple(sorted((num, complement))))
        seen.add(num)

    return len(pairs)

# User input
user_input = input("Enter a list of integers separated by spaces: ")
target_sum = int(input("Enter the target sum: "))
numbers = list(map(int, user_input.strip().split()))
result = count_pairs_with_sum(numbers, target_sum)

print("Number of distinct pairs that sum up to the target:", result)
```

```python
#task6
def remove_duplicates_preserve_order(nums):
    seen = set()
    result = []

    for num in nums:
        if num not in seen:
            seen.add(num)
            result.append(num)

    return result

# User input
user_input = input("Enter a list of integers separated by spaces: ")
numbers = list(map(int, user_input.strip().split()))
result = remove_duplicates_preserve_order(numbers)

print("List after removing duplicates (order preserved):", result)
```

```python
#task7
def get_power_set(nums):
    power_set = [[]]

    for num in nums:
        new_subsets = []
        for subset in power_set:
            new_subsets.append(subset + [num])
        power_set.extend(new_subsets)

    return power_set

# User input
user_input = input("Enter a list of integers separated by spaces: ")
numbers = list(map(int, user_input.strip().split()))
result = get_power_set(numbers)

print("All possible subsets (power set):")
for subset in result:
    print(subset)
```

```python
#task8
def length_of_lis(nums):
    if not nums:
        return 0

    dp = [1] * len(nums)  # Har index pe subsequence length 1 se start hoti hai

    for i in range(len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)

# User input
user_input = input("Enter a list of integers separated by spaces: ")
numbers = list(map(int, user_input.strip().split()))
result = length_of_lis(numbers)

print("Length of the longest increasing subsequence is:", result)
```

```python
#task9
def max_product_of_two(nums):
    if len(nums) < 2:
        return None  # At least 2 numbers required

    nums.sort()
    return max(nums[0] * nums[1], nums[-1] * nums[-2])

# User input
user_input = input("Enter a list of integers separated by spaces: ")
numbers = list(map(int, user_input.strip().split()))
result = max_product_of_two(numbers)

if result is not None:
    print("Maximum product of any two numbers is:", result)
else:
    print("List must contain at least two numbers.")
```

```python
#task13
def top_student(scores_dict):
    if not scores_dict:
        return None  # If dictionary is empty

    return max(scores_dict, key=scores_dict.get)

# User input
students_input = input("Enter student names and scores (e.g. Alice:85 Bob:92 Charlie:88): ")
entries = students_input.strip().split()
students_scores = {}

for entry in entries:
    name, score = entry.split(":")
    students_scores[name] = int(score)

result = top_student(students_scores)

if result:
    print("Top student is:", result)
else:
    print("No students provided.")

```

```python
#task14
def apply_lambda_to_list(nums, func):
    return list(map(func, nums))

# User input
user_input = input("Enter a list of integers separated by spaces: ")
numbers = list(map(int, user_input.strip().split()))

# Lambda function
lambda_func = lambda x: x ** 2  # For squaring each element

result = apply_lambda_to_list(numbers, lambda_func)

print("Output after applying the lambda function:", result)
```

```python
#task15
def calculate(*args, operation='sum'):
    if operation == 'multiply':
        product = 1
        for num in args:
            product *= num
        return product
    else:
        return sum(args)

# Example Usage
result_sum = calculate(1, 2, 3, 4)  # Default operation is 'sum'
result_multiply = calculate(1, 2, 3, 4, operation='multiply')

print("Sum:", result_sum)
print("Product:", result_multiply)
```
